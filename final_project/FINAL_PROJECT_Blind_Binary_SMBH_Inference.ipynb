## Why there are two blind sets

The IID set tests interpolation under the simulation distribution. The shifted set tests whether performance and uncertainty remain defensible when the deployment distribution changes. **A method can score well on IID data and still be scientifically unsafe under shift.**

- **15%** generative problem formulation and leakage control

- **10%** blind numerical performance
- **10%** reproducibility
- **10%** scientific communication / simulation-to-real limitations

Numerical rank is secondary to validity. An overconfident model with slightly lower RMSE can score below a well-calibrated simpler model.
